## 3)Exploratory Data Analysis (EDA) and Visualization
## Campaign Performance:

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

## ➢Plot a scatter plot to visualize the relationship between Acquisition_Cost and ROI

```python
plt.scatter(csv_data["Acquisition_Cost"],csv_data["ROI"],color="lightseagreen",edgecolors="black")
plt.title("Relationship between Acquisition_Cost and ROI")
plt.xlabel("Acquisition_Cost")
plt.ylabel("ROI")
plt.tight_layout()
plt.show()
```

## ➢Create a bar chart to visualize the average Conversion_Rate for different Channel_Used, categorized by Campaign_Type

```python
plt.figure(figsize=(13,5))
bars=csv_data.groupby(["Channel_Used","Campaign_Type"])["Conversion_Rate"].mean().reset_index()
sns.barplot(data=bars,x="Channel_Used",y="Conversion_Rate",hue="Campaign_Type",palette="viridis")
plt.title("Average Conversion_Rate for Different Channel_Used & Categorized by Campaign_Type")
plt.tight_layout()
plt.legend(loc="center")
plt.show()
```

## ➢Visualize the distribution of Engagement_Score across different Campaign_Type using a box plot.

```python
csv_data.boxplot(column="Engagement_Score",by="Campaign_Type",color="blue")
plt.title("Engagement_Score across different Campaign_Type")
plt.ylabel("Engagement_Score")
plt.tight_layout()
plt.show()
```

## ➢Analyze the average ROI by Company using a bar chart to compare the profitability of campaigns conducted by different companies

```python
plt.figure(figsize=(10,5))
sns.barplot(data=csv_data,x="Company",y="ROI",hue="Campaign_Type",estimator="mean",palette="plasma")
plt.title("Analyze the average ROI by Company With Campaigns")
plt.legend(loc="lower center")
plt.tight_layout()
plt.xticks(rotation=45)
plt.show()
```

## ➢Examine the correlation between Engagement_Score and Conversion_Rate using a heatmap

```python
correlation_data=csv_data[["Engagement_Score","Conversion_Rate"]].corr()
sns.heatmap(correlation_data,annot=True,cmap="YlOrRd")
plt.title("Correlation between Engagement_Score and Conversion_Rate")
plt.tight_layout()
plt.show()
```

## Customer Segmentation:
## ➢Create a count plot to visualize the distribution of Target_Audience.

```python
ax=sns.countplot(data=csv_data,x="Target_Audience",color="teal")
ax.bar_label(ax.containers[0])
plt.title("Distribution of Target_Audience")
plt.xticks(rotation=30)
plt.show()
```

## ➢Identify which Customer_Segment has the highest Conversion_Rate for each Language using a bar chart

```python
plt.figure(figsize=(10,6))
sns.barplot(data=csv_data,x="Language",y="Conversion_Rate",hue="Customer_Segment",estimator="max",palette="gist_heat")
plt.title("Highest Conversion_Rate for Each language With Customer_Segement")
plt.legend(loc="center")
plt.tight_layout()
plt.show()
```

## ➢Visualize the distribution of Acquisition_Cost across each Customer_Segment,categorized by Channel_Used, using a box plot

```python
plt.figure(figsize=(10,5))
sns.boxplot(data=csv_data,x="Customer_Segment",y="Acquisition_Cost",hue="Channel_Used",palette="Spectral")
plt.title("Acquisition_Cost across each Customer_Segment by Channel_Used")
plt.legend(loc="best")
plt.tight_layout()
plt.show()
```

## ➢Analyze average Conversion_Rate by Language using a bar chart to compare the effectiveness of campaigns conducted in different languages.

```python
plt.figure(figsize=(10,6))
sns.barplot(data=csv_data,x="Language",y="Conversion_Rate",hue="Campaign_Type",estimator="mean",palette="GnBu")
plt.title("Analyze average Conversion_Rate by Language")
plt.legend(loc="center")
plt.tight_layout()
plt.show()
```

## Channel Effectiveness:
## ➢Compare the Engagement_Score for different Channels_Used, segmented by Campaign_Type, using a bar chart

```python
plt.figure(figsize=(10,6))
sns.barplot(data=csv_data,x="Channel_Used",y="Engagement_Score",hue="Campaign_Type",palette="autumn")
plt.title("Engagement_Score for different Channels_Used, segmented by Campaign_Type")
plt.legend(loc="center")
plt.tight_layout()
plt.show()
```

## ➢Show the distribution of total ROI across different Channels_Used using a pie chart

```python
Total_ROI=csv_data.groupby("Channel_Used")["ROI"].sum().reset_index()
plt.pie(Total_ROI["ROI"],labels=Total_ROI["Channel_Used"],autopct="%1.2f%%",
startangle=30,explode=(0,0.025,0,0,0,0),colors=("green","red","blue","orange","brown","grey"))
plt.title("Distribution of total ROI across different Channels_Used")
plt.tight_layout()
plt.show()
```

## ➢Plot a scatter plot to show the relationship between Clicks and Impressions for each Channel_Used

```python
sns.scatterplot(data=csv_data,x="Clicks",y="Impressions",hue="Channel_Used")
plt.title("Relationship between Clicks and Impressions for each Channel_Used")
plt.show()
```

```python
ROI_Location=csv_data.groupby("Location")["ROI"].sum().reset_index()
plt.pie(data=ROI_Location,x="ROI",labels="Location",autopct="%1.2f%%",startangle=30,explode=(0,0.025,0,0,0,0,0,0,0,0))
plt.title("Proportion of ROI by Location")
plt.tight_layout()
plt.show()
```
